## 1.1.JAVA의 슬로건
- WORA (Write once Run Anywhere) => 운영체제가 무엇이든 똑같은 성능을 낼 수 있다. 
- JVM(Java Virtual Machine) 에 의해 해석되고 실행됨.
- Compile => JVM => OS (a를 0이나 1로 변환시켜서 컴퓨터의 언어로 전달해주는것을 Compile이라고 한다.)

## 1.2.JAVA 개발 환경
- JDK > JRE > JVM
- JDK (Java Development Kit) : 자바 개발 도구 모음
- JRE (Java Runtime Environment) : 자바 실행 환경
- JVM (Java Virtual Machine) : 자바 가상 머신

- IDE (Integrated Development Environment) : 통합 개발 환경
  - 이클립스, 인텔리J, 넷빈즈 등

## 1.4.Java 변수란? 
- Variable(변수) : 변하는 수(열려있는 상자)
- 변수선언 
  - int a = 3; 
    - int : 자료형 (데이터 타입)
    - a : 변수명 (식별자)
    - = : 대입 연산자
    - 3 : 값 (리터럴)

- 상수(Constant) : 변하지 않는 수(닫혀있는 상자)
- 상수선언
  - final int b = 5; 
    - final : 상수 선언 키워드
    - int : 자료형 (데이터 타입)
    - b : 변수명 (식별자)
    - = : 대입 연산자
    - 5 : 값 (리터럴)


## 1.5.변수명 규칙
  - 1.키워드 사용 불가
    - int, double, if, for, while, switch, case, public, private, class 등등   
  - 2.숫자로 시작 불가
    - 숫자로 시작하면 값으로 인식해버린다. compile error 발생
  - 3.특수문자 사용 불가 (_,$ 제외)
    - @, #, %, ^, &, * 등등 사용 불가
  - 4.대소문자 구분
    - Java에서는 변수 이름의 첫 글자는 항상 소문자로 시작하는것이 좋다
    - class 이름은 항상 대문자로 시작하는것이 좋다.
  - 5.의미 있는 이름 사용 권장 (ex. int numStudents;)
  
  - 6.낙타표기법 (camelCase) 권장 (ex. int numStudents;) 
    - 첫 글자는 소문자, 단어가 바뀔때마다 대문자로 시작
  - 기능이 있는 것들은 색을 변환시켜준다.
