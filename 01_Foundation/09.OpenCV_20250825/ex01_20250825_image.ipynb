## 03.1.플래그 설명

| 함수명     | 플래그 속성 명                        | 기능 설명                                                                 |
|------------|--------------------------------------|----------------------------------------------------------------------------|
| imread()   | cv2.IMREAD_COLOR                     | 이미지를 컬러로 로드(투명도는 무시, 기본값)             |
| imread()   | cv2.IMREAD_GRAYSCALE                 | 이미지를 그레이스케일로 로드                                         |
| imread()   | cv2.IMREAD_UNCHANGED                 | 이미지를 알파 채널을 포함하여 로드                                  |
 

```python
# 이미지 불러와서 img 변수에 넣어주기
img = cv2.imread("./data/black_dog.jpg")

# 이미지 출력
plt.imshow(img)
plt.axis('off') # 축 숨기기
plt.show()

# matplotlib : RGB 채널  
# OpenCV : BGR 채널 => R과 B가 바뀌어있다. => 바꿔주면 된다.

# 색상 공간 변환 => BGR to RGB 
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
# plt.axis('off') # 축 숨기기
plt.show()


# 색상 공간 변환 => BGR to GRAY
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(img_gray, cmap='gray')
# plt.axis('off') # 축 숨기기
plt.show()
```

| 함수명     | 플래그 속성 명                        | 기능 설명                                                                 |
|------------|--------------------------------------|----------------------------------------------------------------------------|
| cvtColor() | cv2.COLOR_BGR2GRAY                   | BGR 색상 공간을 그레이스케일로 변환                                  |
| cvtColor() | cv2.COLOR_GRAY2BGR                   | 그레이스케일을 BGR 색상 공간으로 변환                                |

| cvtColor() | cv2.COLOR_RGB2BGR                    | RGB 색상 공간을 BGR 색상 공간으로 변환                               |
| cvtColor() | cv2.COLOR_BGR2HSV                    | BGR 색상 공간을 HSV 색상 공간으로 변환                               |
| cvtColor() | cv2.COLOR_RGB2HSV                    | RGB 색상 공간을 HSV 색상 공간으로 변환                               |
| cvtColor() | cv2.COLOR_HSV2BGR                    | HSV 색상 공간을 BGR 색상 공간으로 변환                               |
| cvtColor() | cv2.COLOR_HSV2RGB                    | HSV 색상 공간을 RGB 색상 공간으로 변환                               |


### 03.2.색상공간 변환
- 시각적인 정보(이미지, 영상 등)들 중에서 컬러는 명확하지 못 한 정보를 보여주는 경우가 있음
- 상황에 따라서 gray 이미지나 binary 이미지로 변환해서 처리
    - gray(회색조) : 픽셀값이 0~255 사이의 값으로 이루어진 흑백 이미지
    - binary(이진화) : 픽셀값이 0(검정색) 또는 255(=1)(흰색)로 구성된 이미지

```python
# 회색조 출력
img = cv2.imread("./data/black_dog.jpg")


print(img.shape) # (높이, 너비, 채널수)
# 색상 공간 변환 => BGR to GRAY
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print(img.shape) # (높이, 너비)
plt.imshow(img_gray, cmap='gray')
# plt.axis('off') # 축 숨기기
plt.show()

# 특정 패턴/물체의 윤곽 감지, 배경/물체 구별, 텍스트 필터링(OCR=광학문자인식), 뼈의 윤곽/종양 식별, 자율주행시 장애물/차선 감지 등 
```

```python
# Binary 이미지 - 이진화(경계값을 기준으로 이진화를 시키면 물체가 뚜렷해짐!)
img = cv2.imread("./data/ryan.jpg", cv2.IMREAD_GRAYSCALE)
img


# 이미지 이진화 시키기 => 2개의 값을 반환시킨다. 
th_val, th_img = cv2.threshold(
    img, # 변환시킬 이미지
    128, # 경계값 => 128을 기준 삼겠다 => 정적으로 지정하면 문제가 생긴다 => 동적으로 조절시켜야겠다 => 경계값 자동 설정 (Otsu's Binarization)
    255, # 최대값 => 255(흰색)로 재할당
    cv2.THRESH_BINARY # 프레그를 이진화 => 픽셀값 > 100 ->255 할당, 100 이하는 0으로 할당
)

th_val # 경계값
th_img # 이진화된 이미지
```

```python
## 원본 이미지
plt.subplot(1, 2, 1) # 여러 그림 그릴 때 자리 지정하는 함수 
plt.axis('off') # 축 숨기기
plt.imshow(img, cmap = "gray")


## 이진화 이미지
plt.subplot(1, 2, 2) # 여러 그림 그릴 때 자리 지정하는 함수 
plt.axis('off') # 축 숨기기
plt.imshow(th_img, cmap = "gray")


plt.show()
```

### 경계값 자동 설정 (Otsu's Binarization)
- Otsu의 이진화: 이미지의 픽셀 분포를 분석해서 최적의 경계값 자동 설정

```python
# 이미지 불러와서 변수에 담아주기
img = cv2.imread("./data/ryan.jpg", cv2.IMREAD_GRAYSCALE )

# Otsu 적용 => 변수에 담아주기
th_val_otsu, th_img_otsu = cv2.threshold(
    img, 
    0, # Otsu 알고리즘이 알아서 찾겠다
    255, 
    cv2.THRESH_BINARY | cv2.THRESH_OTSU # Otsu의 이진화
)
```

```python
## 원본 이미지 => 회색조
plt.subplot(1, 3, 1) # 여러 그림 그릴 때 자리 지정하는 함수 
plt.axis('off') # 축 숨기기
plt.imshow(img, cmap = "gray")


## 이진화 이미지 => 사용자 지정 임계값 기준
plt.subplot(1, 3, 2) # 여러 그림 그릴 때 자리 지정하는 함수 
plt.axis('off') # 축 숨기기
plt.imshow(th_img, cmap = "gray")

## 이진화 이미지 => Otsu
plt.subplot(1, 3, 3) # 여러 그림 그릴 때 자리 지정하는 함수 
plt.axis('off') # 축 숨기기
plt.imshow(th_img_otsu, cmap = "gray")



plt.show()


# 최적 임계값 찾아줘~
th_val_otsu # Otsu가 찾아준 경계값
```

```python
#### 이진화 응용 -사과의 특정 영역에 대한 품질 확인 => 변수에 담아주기
img_apple = cv2.imread("./data/apple.jpg", cv2.IMREAD_GRAYSCALE)
img_apple


# 출력하기 
plt.figure(figsize=(3,3)) # 크기 줄여주기 
plt.imshow(img_apple, cmap='gray') 
plt.show()


# 1.가운데만 상처가 있는지 확인하기(이진화를 통해서 명확하게 파악하기)

## 1.1.특정부분만 잘라내서 가져와서 확인한다 (Resource차원) => ROI(Region of Interest, 관심영역) 
##  이미지의 특정 영역을 확인하기 위해 잘라내기 => X축 y축 4분할 시켜서 가운데 2X2 영역(Region)이  관심영역(Interest)이다.

## 중앙 좌표 기준으로 ROI 수행하기
## 중앙좌표(X, Y) 계산

# 이미지의 크기를 확인한 후, 행과 열 => 높이(height)와 너비(width)를 출력한후 변수에 담아주기
h, w = img_apple.shape

# 중앙좌표(X, Y) 계산
center_x = int(w / 2)
center_y = int(h / 2)

# 중앙좌표값 출력
print(center_x, center_y) 


# ROI 크기지정(중앙의 절반 크기 영역) => 슬라이싱 논리가 적용된다.=> 시작값과 끝값을 지정해야한다. => rect_width, rect_height 변수에 담아주기
rect_width, rect_height = center_x / 2, center_y / 2


# 이미지 슬라이싱 재료 만들어주기 

# 열 시작값 (X축 이라네)
top_left_x = int(center_x - rect_width) # 좌상단 x값
# 행 시작값(y축 이라네)
top_left_y = int(center_y - rect_height) # 좌상단 y값

# 열 끝값
bottom_right_x = int(center_x + rect_width) # 우하단 x값
# 행 끝값
bottom_right_y = int(center_y + rect_height) # 우하단 y값

# 좌표 출력
print(top_left_x, top_left_y) # 좌상단 좌표
print(bottom_right_x, bottom_right_y) # 우하단 좌표


# ROI(Region of Interest) 설정
roi = img_apple[top_left_y:bottom_right_y, top_left_x:bottom_right_x]



# 2.이진화 를 통환 품질확인
val, apple_binary = cv2.threshold(roi, 0, 255, cv2.THRESH_BINARY | cv2.THRESH_OTSU)

plt.figure(figsize=(3,3)) # 크기 줄여주기
plt.imshow(apple_binary, cmap='gray')
plt.show()



# 원본에 합성해서 출력하기
# 원본 이미지 복사
img_apple_copy = img_apple.copy()

# 원본 이미지에 이진화 결과 합성
img_apple_copy[top_left_y:bottom_right_y, top_left_x:bottom_right_x] = apple_binary

# 출력하기 
plt.figure(figsize=(3,3)) # 크기 줄여주기
plt.imshow(img_apple_copy, cmap='gray')
plt.show()
```
