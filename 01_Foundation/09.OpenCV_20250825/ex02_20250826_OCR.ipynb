````python
# PaddleOCR 실제 사용 예제
import cv2
import numpy as np
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt

# 테스트용 한글 이미지 생성
def create_test_image():
    # 흰색 배경에 한글 텍스트가 있는 이미지 생성
    img = Image.new('RGB', (400, 200), color='white')
    draw = ImageDraw.Draw(img)
    
    # 텍스트 추가 (기본 폰트 사용)
    text = "안녕하세요\nPaddleOCR 테스트입니다"
    draw.text((50, 50), text, fill='black')
    
    # 이미지를 numpy 배열로 변환
    img_array = np.array(img)
    return img_array

# 테스트 이미지 생성 및 표시
test_img = create_test_image()
plt.figure(figsize=(8, 4))
plt.imshow(test_img)
plt.title('테스트 이미지')
plt.axis('off')
plt.show()

print("테스트 이미지가 생성되었습니다.")
````

````python
# PaddleOCR로 실제 OCR 수행
try:
    # 위에서 생성한 테스트 이미지를 임시 파일로 저장
    test_image_path = 'temp_test_image.png'
    Image.fromarray(test_img).save(test_image_path)
    
    # OCR 수행
    result = ocr.ocr(test_image_path, cls=True)
    
    print("OCR 결과:")
    for idx, res in enumerate(result):
        print(f"페이지 {idx + 1}:")
        for line in res:
            bbox, (text, confidence) = line
            print(f"  텍스트: '{text}', 신뢰도: {confidence:.4f}")
            print(f"  좌표: {bbox}")
    
    # 결과 시각화
    from paddleocr import draw_ocr
    
    image = Image.open(test_image_path).convert('RGB')
    boxes = [line[0] for line in result[0]]
    texts = [line[1][0] for line in result[0]]
    scores = [line[1][1] for line in result[0]]
    
    # OCR 결과를 이미지에 그리기
    im_show = draw_ocr(image, boxes, texts, scores)
    
    # 결과 표시
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.imshow(test_img)
    plt.title('원본 이미지')
    plt.axis('off')
    
    plt.subplot(1, 2, 2)
    plt.imshow(im_show)
    plt.title('OCR 결과')
    plt.axis('off')
    
    plt.tight_layout()
    plt.show()
    
except Exception as e:
    print(f"OCR 처리 중 오류 발생: {e}")
    print("PaddleOCR가 올바르게 설치되었는지 확인해주세요.")
````

````python
# story.png pytesseract OCR 텍스트 인식 수행
try:
    if os.path.exists('story.png'):
        print("🔍 pytesseract로 텍스트 인식을 시작합니다...")
        print("=" * 60)
        
        # 이미지 로드
        image = Image.open('story.png')
        
        # OCR 수행 (한국어 + 영어)
        try:
            # 한국어 OCR 시도
            text_korean = pytesseract.image_to_string(image, lang='kor')
            print("📖 한국어 OCR 결과:")
            print("=" * 40)
            print(text_korean)
            
            # 상세 정보 추출
            data = pytesseract.image_to_data(image, lang='kor', output_type=pytesseract.Output.DICT)
            
            # 신뢰도가 있는 텍스트만 필터링
            confidences = []
            texts = []
            for i in range(len(data['text'])):
                if int(data['conf'][i]) > 0:  # 신뢰도가 0보다 큰 경우만
                    text = data['text'][i].strip()
                    if text:  # 빈 문자열이 아닌 경우만
                        texts.append(text)
                        confidences.append(int(data['conf'][i]))
            
            print(f"\n� 인식된 단어/구문: {len(texts)}개")
            print("=" * 40)
            
            for i, (text, conf) in enumerate(zip(texts, confidences)):
                print(f"[{i+1:2d}] '{text}' (신뢰도: {conf}%)")
            
            # 텍스트 분석
            full_text = text_korean.strip()
            if full_text:
                print("\n📈 텍스트 분석:")
                print("=" * 40)
                print(f"• 총 글자 수: {len(full_text)}자")
                print(f"• 줄 수: {len(full_text.split())}")
                
                # 한글 비율 계산
                korean_chars = sum(1 for char in full_text if '\uac00' <= char <= '\ud7af')
                if len(full_text) > 0:
                    korean_ratio = korean_chars / len(full_text) * 100
                    print(f"• 한글 비율: {korean_ratio:.1f}% ({korean_chars}/{len(full_text)})")
                
                # 평균 신뢰도
                if confidences:
                    avg_conf = np.mean(confidences)
                    print(f"• 평균 신뢰도: {avg_conf:.1f}%")
            else:
                print("❌ 텍스트를 인식하지 못했습니다.")
                
        except Exception as tesseract_error:
            print(f"❌ Tesseract OCR 오류: {tesseract_error}")
            print("Tesseract 실행파일이 설치되어 있지 않을 수 있습니다.")
            print("Windows: https://github.com/UB-Mannheim/tesseract/wiki 에서 다운로드")
            
            # 대안으로 OpenCV를 이용한 기본 이미지 분석
            print("\n🔍 대안: OpenCV를 이용한 기본 이미지 분석")
            img_cv = cv2.imread('story.png')
            if img_cv is not None:
                height, width = img_cv.shape[:2]
                print(f"• 이미지 크기: {width} x {height} 픽셀")
                
                # 그레이스케일 변환
                gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
                
                # 텍스트 영역 감지 (윤곽선 기반)
                _, thresh = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
                contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
                
                print(f"• 감지된 영역: {len(contours)}개")
                print("• OCR 없이는 텍스트 내용을 읽을 수 없습니다.")
            
    else:
        print("❌ story.png 파일을 찾을 수 없어 OCR을 수행할 수 없습니다.")
        print("현재 디렉토리의 이미지 파일들:")
        for file in os.listdir('.'):
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                print(f"  - {file}")
        
except Exception as e:
    print(f"OCR 처리 중 오류 발생: {e}")
````

````python
# 이미지 분석 및 텍스트 영역 시각화 (OCR 결과 없이도 가능)
try:
    if os.path.exists('story.png'):
        print("🎨 이미지 분석 및 시각화를 시작합니다...")
        
        # 이미지 로드
        image = cv2.imread('story.png')
        image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
        
        # 이미지 전처리 및 텍스트 영역 감지
        # 1. 적응적 임계값을 사용한 이진화
        thresh = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, 
                                      cv2.THRESH_BINARY, 11, 2)
        
        # 2. 윤곽선 검출
        contours, _ = cv2.findContours(thresh, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        
        # 3. 텍스트 영역으로 추정되는 윤곽선 필터링
        text_contours = []
        for contour in contours:
            area = cv2.contourArea(contour)
            if area > 100:  # 너무 작은 영역 제외
                x, y, w, h = cv2.boundingRect(contour)
                aspect_ratio = w / h
                if 0.1 < aspect_ratio < 10:  # 적절한 가로세로 비율
                    text_contours.append(contour)
        
        # 시각화
        plt.figure(figsize=(16, 12))
        
        # 원본 이미지
        plt.subplot(2, 2, 1)
        plt.imshow(image_rgb)
        plt.title('story.png - 원본 이미지', fontsize=14, fontweight='bold')
        plt.axis('off')
        
        # 그레이스케일
        plt.subplot(2, 2, 2)
        plt.imshow(gray, cmap='gray')
        plt.title('그레이스케일 변환', fontsize=14, fontweight='bold')
        plt.axis('off')
        
        # 이진화
        plt.subplot(2, 2, 3)
        plt.imshow(thresh, cmap='gray')
        plt.title('적응적 임계값 이진화', fontsize=14, fontweight='bold')
        plt.axis('off')
        
        # 텍스트 영역 표시
        img_contours = image_rgb.copy()
        for i, contour in enumerate(text_contours):
            # 바운딩 박스 그리기
            x, y, w, h = cv2.boundingRect(contour)
            cv2.rectangle(img_contours, (x, y), (x+w, y+h), (0, 255, 0), 2)
            cv2.putText(img_contours, f'{i+1}', (x, y-5), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 0, 0), 1)
        
        plt.subplot(2, 2, 4)
        plt.imshow(img_contours)
        plt.title(f'텍스트 영역 추정 ({len(text_contours)}개 영역)', fontsize=14, fontweight='bold')
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()
        
        # 이미지 통계
        print("📊 이미지 분석 결과:")
        print("=" * 50)
        print(f"• 이미지 크기: {image_rgb.shape[1]} x {image_rgb.shape[0]} 픽셀")
        print(f"• 색상 채널: {image_rgb.shape[2]}개")
        print(f"• 추정된 텍스트 영역: {len(text_contours)}개")
        
        # 영역별 정보
        if text_contours:
            areas = [cv2.contourArea(contour) for contour in text_contours]
            print(f"• 평균 영역 크기: {np.mean(areas):.0f} 픽셀²")
            print(f"• 최대 영역 크기: {np.max(areas):.0f} 픽셀²")
            print(f"• 최소 영역 크기: {np.min(areas):.0f} 픽셀²")
        
        print("\n💡 OCR 텍스트 인식을 위해서는 Tesseract가 필요합니다.")
        print("   위 분석은 이미지의 텍스트 영역만 추정한 것입니다.")
        
    else:
        print("❌ story.png 파일을 찾을 수 없습니다.")
        
except Exception as e:
    print(f"이미지 분석 중 오류 발생: {e}")
````

````python
# 3단계: 번호판이 위치한 영역에 대해 ROI 설정
def detect_license_plate_region(image):
    """
    자동차 번호판 영역을 자동으로 탐지하는 함수
    """
    # 그레이스케일 변환
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    
    # 노이즈 제거를 위한 블러링
    blurred = cv2.GaussianBlur(gray, (5, 5), 0)
    
    # 에지 검출
    edges = cv2.Canny(blurred, 50, 150)
    
    # 윤곽선 검출
    contours, _ = cv2.findContours(edges, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    # 번호판 후보 영역 필터링
    license_plate_candidates = []
    
    for contour in contours:
        area = cv2.contourArea(contour)
        if area > 1000:  # 최소 면적 조건
            x, y, w, h = cv2.boundingRect(contour)
            aspect_ratio = w / h
            
            # 번호판의 일반적인 가로세로 비율 (2:1 ~ 5:1)
            if 2.0 <= aspect_ratio <= 5.0 and w > 50 and h > 15:
                license_plate_candidates.append((x, y, w, h, area))
    
    # 면적 기준으로 정렬하여 가장 큰 후보 선택
    if license_plate_candidates:
        license_plate_candidates.sort(key=lambda x: x[4], reverse=True)
        return license_plate_candidates[:3]  # 상위 3개 후보 반환
    else:
        return []

# car.png 이미지가 로드되었다면 번호판 영역 탐지
if 'car_image' in locals():
    print("🔍 번호판 영역을 탐지하는 중...")
    
    plate_candidates = detect_license_plate_region(car_image)
    
    if plate_candidates:
        print(f"✅ {len(plate_candidates)}개의 번호판 후보 영역을 발견했습니다.")
        
        # 결과 시각화
        img_with_boxes = car_image_rgb.copy()
        
        plt.figure(figsize=(15, 10))
        
        # 원본 이미지
        plt.subplot(2, 2, 1)
        plt.imshow(car_image_rgb)
        plt.title('원본 자동차 이미지', fontsize=14)
        plt.axis('off')
        
        # 에지 검출 결과
        gray = cv2.cvtColor(car_image, cv2.COLOR_BGR2GRAY)
        blurred = cv2.GaussianBlur(gray, (5, 5), 0)
        edges = cv2.Canny(blurred, 50, 150)
        
        plt.subplot(2, 2, 2)
        plt.imshow(edges, cmap='gray')
        plt.title('에지 검출 결과', fontsize=14)
        plt.axis('off')
        
        # 번호판 후보 영역 표시
        for i, (x, y, w, h, area) in enumerate(plate_candidates):
            cv2.rectangle(img_with_boxes, (x, y), (x+w, y+h), (255, 0, 0), 3)
            cv2.putText(img_with_boxes, f'후보 {i+1}', (x, y-10), 
                       cv2.FONT_HERSHEY_SIMPLEX, 0.7, (255, 0, 0), 2)
        
        plt.subplot(2, 2, 3)
        plt.imshow(img_with_boxes)
        plt.title('번호판 후보 영역', fontsize=14)
        plt.axis('off')
        
        # 가장 유력한 후보 영역 확대
        best_x, best_y, best_w, best_h, _ = plate_candidates[0]
        roi = car_image_rgb[best_y:best_y+best_h, best_x:best_x+best_w]
        
        plt.subplot(2, 2, 4)
        plt.imshow(roi)
        plt.title('ROI - 번호판 추정 영역', fontsize=14)
        plt.axis('off')
        
        plt.tight_layout()
        plt.show()
        
        # ROI 좌표 정보 출력
        print("\n📊 탐지된 번호판 후보 영역:")
        for i, (x, y, w, h, area) in enumerate(plate_candidates):
            print(f"후보 {i+1}: 위치({x}, {y}), 크기({w}x{h}), 면적({area:.0f})")
        
        # 최적 ROI 저장
        best_roi_coords = (best_x, best_y, best_w, best_h)
        print(f"\n🎯 선택된 ROI: {best_roi_coords}")
        
    else:
        print("❌ 번호판 영역을 자동으로 탐지하지 못했습니다.")
        print("수동으로 ROI를 설정해야 할 수 있습니다.")
        
        # 수동 ROI 설정 예시 (이미지 크기에 따라 조정 필요)
        height, width = car_image.shape[:2]
        manual_x = width // 4
        manual_y = height // 2
        manual_w = width // 2
        manual_h = height // 4
        
        best_roi_coords = (manual_x, manual_y, manual_w, manual_h)
        print(f"🔧 수동 ROI 설정: {best_roi_coords}")
else:
    print("❌ 이미지가 로드되지 않았습니다. 1단계를 먼저 실행해주세요.")
````

````python
# 4단계: ROI 수행 및 이미지 전처리 (문자 영역만 남기도록 ROI 정제 포함)
def refine_roi_to_characters(roi_bgr, padding_ratio: float = 0.03):
    """
    번호판 ROI에서 숫자/한글 문자 라인만 남기도록 자동으로 내부 영역을 타이트하게 크롭
    - 문자 연결을 위해 가로 방향 중심의 모폴로지 후 외곽 윤곽을 합쳐 bbox 산출
    """
    gray = cv2.cvtColor(roi_bgr, cv2.COLOR_BGR2GRAY)
    gray_blur = cv2.GaussianBlur(gray, (3, 3), 0)

    # 문자(검정)를 전경으로 만들기 위해 THRESH_BINARY_INV 사용
    bin_inv = cv2.adaptiveThreshold(
        gray_blur, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY_INV,
        31, 15
    )

    h, w = bin_inv.shape[:2]
    # 가로로 문자를 연결하기 위한 커널 (이미지 크기에 비례)
    kernel = cv2.getStructuringElement(
        cv2.MORPH_RECT, (max(5, w // 15), max(1, h // 8))
    )
    mor = cv2.morphologyEx(bin_inv, cv2.MORPH_CLOSE, kernel, iterations=1)

    # 외곽 윤곽 추출
    contours, _ = cv2.findContours(mor, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

    boxes = []
    area_min = (h * w) * 0.01  # 너무 작은 잡음 제거
    for c in contours:
        x, y, ww, hh = cv2.boundingRect(c)
        area = ww * hh
        if area < area_min:
            continue
        aspect = ww / float(hh)
        # 과도하게 납작/가늘거나 너무 낮은 박스 제외, 문자의 높이가 충분한 것만 사용
        if 0.2 < aspect < 15 and hh > h * 0.35:
            boxes.append((x, y, ww, hh))

    if boxes:
        x0 = min(b[0] for b in boxes)
        y0 = min(b[1] for b in boxes)
        x1 = max(b[0] + b[2] for b in boxes)
        y1 = max(b[1] + b[3] for b in boxes)
    else:
        # 실패 시 안전한 기본 크롭(여백 제거)으로 폴백
        x0, x1 = int(w * 0.08), int(w * 0.92)
        y0, y1 = int(h * 0.25), int(h * 0.85)

    # 소량 패딩 후 범위 클램프
    pad_x = int((x1 - x0) * padding_ratio)
    pad_y = int((y1 - y0) * padding_ratio)
    x0 = max(0, x0 - pad_x)
    y0 = max(0, y0 - pad_y)
    x1 = min(w, x1 + pad_x)
    y1 = min(h, y1 + pad_y)

    refined = roi_bgr[y0:y1, x0:x1].copy()
    debug = {
        'gray': gray,
        'binary_inv': bin_inv,
        'morph': mor,
        'refined_rect_local': (x0, y0, x1 - x0, y1 - y0)
    }
    return refined, debug


def preprocess_license_plate(roi_image):
    """
    번호판 ROI 이미지를 OCR에 최적화된 형태로 전처리
    """
    # 그레이스케일 변환
    gray = cv2.cvtColor(roi_image, cv2.COLOR_BGR2GRAY)

    # 노이즈 제거
    denoised = cv2.medianBlur(gray, 3)

    # 히스토그램 평활화로 대비 개선
    enhanced = cv2.equalizeHist(denoised)

    # 적응적 이진화 (텍스트 인식 향상)
    binary = cv2.adaptiveThreshold(
        enhanced, 255, cv2.ADAPTIVE_THRESH_GAUSSIAN_C, cv2.THRESH_BINARY, 11, 2
    )

    # 모폴로지 연산으로 텍스트 정리
    kernel = cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2))
    cleaned = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)

    # 이미지 크기 조정 (OCR 정확도 향상을 위해)
    height, width = cleaned.shape
    scale_factor = max(2, 200 // max(1, height))  # 최소 높이 200px 확보
    new_width = int(width * scale_factor)
    new_height = int(height * scale_factor)

    resized = cv2.resize(cleaned, (new_width, new_height), interpolation=cv2.INTER_CUBIC)

    return resized, {
        'gray': gray,
        'denoised': denoised,
        'enhanced': enhanced,
        'binary': binary,
        'cleaned': cleaned
    }


# ROI가 설정되었다면 전처리 수행
if 'best_roi_coords' in locals() and 'car_image' in locals():
    print("🔧 ROI 영역을 추출하고(문자 영역으로 정제) 전처리하는 중...")

    # 1) 원본 ROI 추출
    x, y, w, h = best_roi_coords
    roi_bgr_full = car_image[y:y+h, x:x+w]

    # 2) 문자 라인만 남기도록 ROI 정제
    roi_bgr, refine_dbg = refine_roi_to_characters(roi_bgr_full, padding_ratio=0.04)
    rx, ry, rw, rh = refine_dbg['refined_rect_local']  # ROI 내부 좌표계

    # 시각화용 RGB
    roi_rgb = cv2.cvtColor(roi_bgr, cv2.COLOR_BGR2RGB)

    # 3) 전처리 수행
    processed_roi, processing_steps = preprocess_license_plate(roi_bgr)

    print("✅ ROI 전처리가 완료되었습니다.")

    # 전처리 단계별 결과 시각화
    plt.figure(figsize=(18, 12))

    # (1) 정제 전/후 ROI 비교
    plt.subplot(3, 3, 1)
    plt.imshow(cv2.cvtColor(roi_bgr_full, cv2.COLOR_BGR2RGB))
    plt.title('원본 ROI (번호판 전체)', fontsize=12)
    plt.axis('off')

    plt.subplot(3, 3, 2)
    plt.imshow(refine_dbg['binary_inv'], cmap='gray')
    plt.title('문자 검출용 이진화(반전)', fontsize=12)
    plt.axis('off')

    plt.subplot(3, 3, 3)
    plt.imshow(roi_rgb)
    plt.title('정제된 ROI (문자 영역)', fontsize=12)
    plt.axis('off')

    # (2) 전처리 파이프라인
    plt.subplot(3, 3, 4)
    plt.imshow(processing_steps['gray'], cmap='gray')
    plt.title('1. 그레이스케일', fontsize=12)
    plt.axis('off')

    plt.subplot(3, 3, 5)
    plt.imshow(processing_steps['denoised'], cmap='gray')
    plt.title('2. 노이즈 제거', fontsize=12)
    plt.axis('off')

    plt.subplot(3, 3, 6)
    plt.imshow(processing_steps['enhanced'], cmap='gray')
    plt.title('3. 대비 개선', fontsize=12)
    plt.axis('off')

    plt.subplot(3, 3, 7)
    plt.imshow(processing_steps['binary'], cmap='gray')
    plt.title('4. 적응적 이진화', fontsize=12)
    plt.axis('off')

    plt.subplot(3, 3, 8)
    plt.imshow(processing_steps['cleaned'], cmap='gray')
    plt.title('5. 모폴로지 정리', fontsize=12)
    plt.axis('off')

    # 최종 처리된 이미지
    plt.subplot(3, 3, 9)
    plt.imshow(processed_roi, cmap='gray')
    plt.title('6. 최종 처리 결과 (크기 조정)', fontsize=12)
    plt.axis('off')

    plt.tight_layout()
    plt.show()

    # ROI 위치 정보(전체 이미지에 표시: 빨강=원본 ROI, 초록=정제 ROI)
    full_img_with_roi = car_image_rgb.copy()
    cv2.rectangle(full_img_with_roi, (x, y), (x+w, y+h), (255, 0, 0), 3)  # 원본 ROI
    cv2.rectangle(full_img_with_roi, (x+rx, y+ry), (x+rx+rw, y+ry+rh), (0, 255, 0), 3)  # 정제 ROI

    plt.figure(figsize=(8, 5))
    plt.imshow(full_img_with_roi)
    plt.title('전체 이미지에서 ROI (빨강: 원본, 초록: 문자영역)', fontsize=12)
    plt.axis('off')
    plt.show()

    # ROI 정보 출력
    print(f"\n📏 원본 ROI 크기: {roi_bgr_full.shape} / 좌표: x={x}, y={y}, w={w}, h={h}")
    print(f"📏 정제 ROI 크기: {roi_rgb.shape} / 좌표(전체 기준): x={x+rx}, y={y+ry}, w={rw}, h={rh}")
else:
    print("❌ ROI가 설정되지 않았습니다. 3단계를 먼저 실행해주세요.")
````

````python
# 5단계: OCR 수행 (강화 버전)
import re

def _build_ocr_variants(img):
    # 그레이스케일 보장
    if len(img.shape) == 3:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    else:
        gray = img.copy()

    # 검정 글자/흰 배경 보장 (필요시 반전)
    black_ratio = (gray < 128).mean()
    base = gray if black_ratio > 0.5 else cv2.bitwise_not(gray)

    # 최소 높이 320으로 확대
    h, w = base.shape
    scale = max(1, 320 // max(1, h))
    up = cv2.resize(base, (w*scale, h*scale), interpolation=cv2.INTER_CUBIC)

    variants = [
        ("base", up),
        ("median", cv2.medianBlur(up, 3)),
        ("bilateral", cv2.bilateralFilter(up, 7, 40, 40)),
        ("dilate", cv2.dilate(up, cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2)), 1)),
        ("erode",  cv2.erode(up,  cv2.getStructuringElement(cv2.MORPH_RECT, (2, 2)), 1)),
    ]

    # 오츠 이진화로 재정리
    cleaned = []
    for name, v in variants:
        v2 = cv2.threshold(v, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
        cleaned.append((name, v2))
    return cleaned

def _clean_text(text: str) -> str:
    # 숫자/영문/한글만 남김
    return ''.join(re.findall(r'[0-9A-Za-z가-힣]+', text))

def _calc_confidence(conf_list):
    vals = []
    for c in conf_list:
        try:
            v = float(c)
            if v > 0:
                vals.append(v)
        except:
            continue
    return (sum(vals) / len(vals)) if vals else 0.0

def _pattern_score(text: str) -> float:
    # 한국 번호판(2숫자+한글+4숫자)에 가중치
    text_nospace = re.sub(r'\s+', '', text)
    bonus = 0.0
    if re.fullmatch(r'[0-9]{2}[가-힣][0-9]{4}', text_nospace):
        bonus = 25.0
    elif re.search(r'[0-9]{2}.*[가-힣].*[0-9]{3,4}', text):
        bonus = 15.0
    # 글자수 유사도(목표 7자)
    bonus += min(len(text_nospace), 7) / 7 * 10.0
    return bonus

def perform_license_plate_ocr(processed_image):
    """
    전처리된 번호판 이미지에서 텍스트를 추출 (강화: 다중 전처리/PSM/엔진, 패턴 가중치)
    """
    try:
        whitelist = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ가나다라마바사아자차카타파하허호서울부산대구인천광주대전울산세종경기강원충북충남전북전남경북경남제주"
        base_cfg = f"--oem 3 -l kor+eng -c tessedit_char_whitelist={whitelist}"
        psm_list = [7, 8, 6, 13]  # 단일라인, 단어, 단락, 원시줄
        configs = [f"{base_cfg} --psm {p}" for p in psm_list]

        results = []
        variants = _build_ocr_variants(processed_image)

        for vname, vimg in variants:
            for i, config in enumerate(configs, start=1):
                text = pytesseract.image_to_string(vimg, config=config)
                data = pytesseract.image_to_data(vimg, config=config, output_type=pytesseract.Output.DICT)
                avg_conf = _calc_confidence(data.get('conf', []))
                cleaned = _clean_text(text)

                results.append({
                    "variant": vname,
                    "config_id": i,
                    "config": config,
                    "raw_text": text,
                    "cleaned_text": cleaned,
                    "confidence": avg_conf,
                    "score": avg_conf + _pattern_score(cleaned),
                    "char_count": len(cleaned),
                })

        # 보조 엔진(PaddleOCR) 사용 가능 시 추가 시도
        if 'ocr' in globals() or 'ocr' in locals():
            try:
                paddle_res = (ocr.ocr(cv2.cvtColor(processed_image, cv2.COLOR_GRAY2RGB) if len(processed_image.shape)==2 else processed_image, cls=True) or [[]])[0]
                paddle_text = _clean_text(''.join([r[1][0] for r in paddle_res]))
                paddle_conf = np.mean([r[1][1]*100 for r in paddle_res]) if paddle_res else 0.0
                results.append({
                    "variant": "paddle",
                    "config_id": 0,
                    "config": "PaddleOCR(korean)",
                    "raw_text": paddle_text,
                    "cleaned_text": paddle_text,
                    "confidence": paddle_conf,
                    "score": paddle_conf + _pattern_score(paddle_text),
                    "char_count": len(paddle_text),
                })
            except Exception:
                pass

        return results
    except Exception as e:
        return [{"error": f"OCR 실행 중 오류 발생: {str(e)}"}]

# 전처리된 이미지가 있다면 OCR 수행
if 'processed_roi' in locals():
    plt.rcParams['font.family'] = 'Malgun Gothic'
    plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지   

    print("🔤 OCR을 수행하는 중(강화 모드)...")

    ocr_results = perform_license_plate_ocr(processed_roi)

    print("✅ OCR 수행이 완료되었습니다.")

    # 결과 시각화
    plt.figure(figsize=(16, 10))

    # 처리된 이미지 표시
    plt.subplot(2, 3, 1)
    plt.imshow(processed_roi, cmap='gray')
    plt.title('OCR 대상 이미지', fontsize=14)
    plt.axis('off')

    # 원본 ROI와 비교
    plt.subplot(2, 3, 2)
    plt.imshow(roi_rgb)
    plt.title('원본 ROI', fontsize=14)
    plt.axis('off')

    # OCR 결과 요약
    plt.subplot(2, 3, (3, 6))
    plt.axis('off')
    plt.rcParams['font.family'] = 'Malgun Gothic'
    plt.rcParams['axes.unicode_minus'] = False  # 마이너스 기호 깨짐 방지

    result_text = "🔍 OCR 분석 결과(강화)\n" + "="*50 + "\n\n"

    valid_results = [r for r in ocr_results if 'error' not in r]

    if valid_results:
        # 점수(신뢰도+패턴) 기준으로 정렬
        valid_results.sort(key=lambda x: x['score'], reverse=True)

        result_text += f"📊 총 {len(valid_results)}개 시도(전처리×설정) 수행\n\n"

        for r in valid_results[:8]:
            result_text += f"[{r['variant']}/설정{r['config_id']}] \n"
            result_text += f"  원본: '{r['raw_text'].strip()}'\n"
            result_text += f"  정리: '{r['cleaned_text']}'\n"
            result_text += f"  신뢰도: {r['confidence']:.1f}% | 점수: {r['score']:.1f}\n\n"

        best_result = valid_results[0]
        result_text += "🏆 최고 결과:\n"
        result_text += f"  번호판 텍스트: '{best_result['cleaned_text']}'\n"
        result_text += f"  신뢰도: {best_result['confidence']:.1f}% (변형: {best_result['variant']}, 설정: {best_result['config_id']})\n"

        final_license_plate_text = best_result['cleaned_text']
        final_confidence = best_result['confidence']
    else:
        result_text += "❌ 모든 OCR 설정에서 텍스트를 인식하지 못했습니다.\n"
        result_text += "이미지 품질이나 전처리 과정을 개선해야 할 수 있습니다."
        final_license_plate_text = ""
        final_confidence = 0

    plt.text(0.05, 0.95, result_text, transform=plt.gca().transAxes,
             fontsize=11, verticalalignment='top', fontfamily='monospace',
             bbox=dict(boxstyle="round,pad=0.5", facecolor="lightblue", alpha=0.8))

    # 상위 결과 점수 막대 그래프
    plt.subplot(2, 3, 4)
    if valid_results:
        labels = [f"{r['variant']}-설정{r['config_id']}" for r in valid_results[:8]]
        scores = [r['score'] for r in valid_results[:8]]
        plt.barh(labels, scores, color='skyblue', alpha=0.8)
        plt.title('상위 시도 점수(신뢰도+패턴)', fontsize=12)
        plt.gca().invert_yaxis()
        plt.grid(True, alpha=0.3)
    else:
        plt.text(0.5, 0.5, 'OCR 결과 없음', ha='center', va='center', transform=plt.gca().transAxes)
        plt.title('점수 분석')

    # 문자 길이 분석
    plt.subplot(2, 3, 5)
    if valid_results:
        char_counts = [r['char_count'] for r in valid_results[:8]]
        plt.bar(range(len(char_counts)), char_counts, color='lightcoral', alpha=0.7)
        plt.title('인식된 문자 수', fontsize=12)
        plt.ylabel('문자 개수')
        plt.xticks(range(len(char_counts)), [f"{i+1}" for i in range(len(char_counts))])
        plt.grid(True, alpha=0.3)
    else:
        plt.text(0.5, 0.5, '문자 인식 없음', ha='center', va='center', transform=plt.gca().transAxes)
        plt.title('문자 수 분석')

    plt.tight_layout()
    plt.show()

    print("\n" + "="*60)
    print("🏁 OCR 수행 완료(강화)")
    print("="*60)

    if valid_results:
        print(f"✅ 최종 인식 결과: '{final_license_plate_text}'")
        print(f"📊 신뢰도: {final_confidence:.1f}%")
        print(f"📝 문자 수: {len(final_license_plate_text)}")
    else:
        print("❌ 번호판 텍스트를 인식하지 못했습니다.")
else:
    print("❌ 전처리된 이미지가 없습니다. 4단계를 먼저 실행해주세요.")
````

````python
# 6단계: OCR로 추출된 텍스트 출력 및 최종 결과 정리
def analyze_license_plate_text(text, confidence):
    """
    번호판 텍스트를 분석하고 유효성을 검사
    """
    analysis = {
        'original_text': text,
        'length': len(text),
        'is_valid': False,
        'plate_type': 'Unknown',
        'region': 'Unknown',
        'numbers': '',
        'letters': '',
        'korean_chars': ''
    }
    
    if not text:
        return analysis
    
    # 문자 분류
    numbers = ''.join([c for c in text if c.isdigit()])
    letters = ''.join([c for c in text if c.isalpha() and ord(c) < 128])
    korean_chars = ''.join([c for c in text if ord(c) > 127 and c.isalpha()])
    
    analysis['numbers'] = numbers
    analysis['letters'] = letters
    analysis['korean_chars'] = korean_chars
    
    # 한국 번호판 패턴 분석
    korean_regions = ['서울', '부산', '대구', '인천', '광주', '대전', '울산', '세종', 
                     '경기', '강원', '충북', '충남', '전북', '전남', '경북', '경남', '제주']
    
    # 지역명 검출
    for region in korean_regions:
        if region in text:
            analysis['region'] = region
            break
    
    # 번호판 유형 판단
    if len(text) >= 7 and len(numbers) >= 4:
        if any(region in text for region in korean_regions):
            analysis['plate_type'] = '일반 번호판'
            analysis['is_valid'] = True
        elif len(numbers) >= 4:
            analysis['plate_type'] = '추정 번호판'
            analysis['is_valid'] = True
    
    return analysis

def format_final_output(text, confidence, analysis):
    """
    최종 결과를 깔끔하게 포맷팅
    """
    output = []
    output.append("🚗 자동차 번호판 OCR 분석 최종 결과")
    output.append("=" * 50)
    output.append("")
    
    if text:
        output.append(f"🔤 인식된 텍스트: '{text}'")
        output.append(f"📊 신뢰도: {confidence:.1f}%")
        output.append(f"📏 문자 길이: {analysis['length']}자")
        output.append("")
        
        output.append("📋 텍스트 구성 요소:")
        output.append(f"  • 숫자: '{analysis['numbers']}' ({len(analysis['numbers'])}개)")
        output.append(f"  • 영문: '{analysis['letters']}' ({len(analysis['letters'])}개)")
        output.append(f"  • 한글: '{analysis['korean_chars']}' ({len(analysis['korean_chars'])}개)")
        output.append("")
        
        output.append("🏷️ 번호판 분석:")
        output.append(f"  • 유형: {analysis['plate_type']}")
        output.append(f"  • 지역: {analysis['region']}")
        output.append(f"  • 유효성: {'✅ 유효' if analysis['is_valid'] else '❌ 무효'}")
        output.append("")
        
        # 신뢰도에 따른 권고사항
        if confidence >= 80:
            output.append("💯 높은 신뢰도 - 인식 결과가 매우 정확할 가능성이 높습니다.")
        elif confidence >= 60:
            output.append("👍 중간 신뢰도 - 인식 결과가 대체로 정확합니다.")
        elif confidence >= 40:
            output.append("⚠️ 낮은 신뢰도 - 결과 검증이 필요합니다.")
        else:
            output.append("❌ 매우 낮은 신뢰도 - 이미지 품질 개선이 필요합니다.")
            
    else:
        output.append("❌ 텍스트를 인식하지 못했습니다.")
        output.append("")
        output.append("💡 개선 방안:")
        output.append("  • 이미지 해상도 향상")
        output.append("  • 조명 개선")
        output.append("  • 번호판 각도 조정")
        output.append("  • ROI 영역 재설정")
    
    return "\n".join(output)

# 최종 결과 출력
if 'final_license_plate_text' in locals() and 'final_confidence' in locals():
    print("📄 최종 결과를 정리하는 중...")
    
    # 텍스트 분석
    text_analysis = analyze_license_plate_text(final_license_plate_text, final_confidence)
    
    # 최종 출력 포맷팅
    final_output = format_final_output(final_license_plate_text, final_confidence, text_analysis)
    
    # 시각적 결과 표시
    plt.figure(figsize=(20, 12))
    
    # 1. 전체 프로세스 요약
    plt.subplot(2, 4, 1)
    plt.imshow(car_image_rgb)
    plt.title('1단계: 원본 이미지', fontsize=12)
    plt.axis('off')
    
    # 2. ROI 영역
    plt.subplot(2, 4, 2)
    roi_with_box = car_image_rgb.copy()
    if 'best_roi_coords' in locals():
        x, y, w, h = best_roi_coords
        cv2.rectangle(roi_with_box, (x, y), (x+w, y+h), (255, 0, 0), 3)
    plt.imshow(roi_with_box)
    plt.title('2단계: ROI 설정', fontsize=12)
    plt.axis('off')
    
    # 3. 추출된 ROI
    if 'roi_rgb' in locals():
        plt.subplot(2, 4, 3)
        plt.imshow(roi_rgb)
        plt.title('3단계: ROI 추출', fontsize=12)
        plt.axis('off')
    
    # 4. 전처리된 이미지
    if 'processed_roi' in locals():
        plt.subplot(2, 4, 4)
        plt.imshow(processed_roi, cmap='gray')
        plt.title('4단계: 이미지 전처리', fontsize=12)
        plt.axis('off')
    
    # 5-8. 최종 결과 텍스트
    plt.subplot(2, 4, (5, 8))
    plt.axis('off')
    
    # 결과를 색상별로 구분하여 표시
    if final_license_plate_text:
        bg_color = 'lightgreen' if final_confidence >= 60 else 'lightyellow' if final_confidence >= 40 else 'lightcoral'
    else:
        bg_color = 'lightgray'
    
    plt.text(0.05, 0.95, final_output, transform=plt.gca().transAxes,
             fontsize=11, verticalalignment='top', fontfamily='monospace',
             bbox=dict(boxstyle="round,pad=0.5", facecolor=bg_color, alpha=0.8))
    
    plt.suptitle('🚗 자동차 번호판 OCR 전체 프로세스 결과', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.show()
    
    # 콘솔에 최종 결과 출력
    print("\n" + final_output)
    
    # 추가 정보
    print("\n" + "="*50)
    print("📊 처리 과정 요약:")
    print(f"✅ 1단계: 이미지 로드 완료")
    print(f"✅ 2단계: OCR 초기화 완료")
    print(f"✅ 3단계: ROI 영역 탐지 완료")
    print(f"✅ 4단계: 이미지 전처리 완료")
    print(f"✅ 5단계: OCR 수행 완료")
    print(f"✅ 6단계: 결과 분석 완료")
    
    if 'best_roi_coords' in locals():
        x, y, w, h = best_roi_coords
        print(f"\n🎯 최종 ROI 좌표: ({x}, {y}, {w}, {h})")
    
    if 'ocr_results' in locals():
        valid_count = len([r for r in ocr_results if 'error' not in r])
        print(f"🔍 OCR 시도 횟수: {len(ocr_results)}번 중 {valid_count}번 성공")
    
    print("\n🎉 자동차 번호판 OCR 분석이 완료되었습니다!")
    
else:
    print("❌ OCR 결과가 없습니다. 5단계를 먼저 실행해주세요.")
    print("\n📋 전체 프로세스:")
    print("1. 이미지 로드")
    print("2. OCR 초기화") 
    print("3. ROI 설정")
    print("4. 이미지 전처리")
    print("5. OCR 수행")
    print("6. 결과 출력 ← 현재 단계")
    print("\n위 단계들을 순서대로 실행해주세요.")
````
