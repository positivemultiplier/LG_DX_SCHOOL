## 1.4.머신러닝 이전에는 어떤 방식으로 동작을 자동화 했을까?
### 1.4.1. Herbert A. Simon은 제한된 합리성(Bounded Rationality) 문제를 해결하기 위해 **Carnegie Mellon University에서 인공지능 연구소(Artificial Intelligence Laboratory)**를 설립했습니다.[1][2]

### 1.4.2.  핵심 조직 설립

Simon은 Allen Newell과 함께 **Carnegie Mellon University의 인공지능 연구소**를 설립하여 제한된 합리성 문제에 대한 해결책을 모색했습니다. 이 연구소에서 그들은 인간의 문제 해결 과정을 컴퓨터로 시뮬레이션하는 연구를 진행했습니다.[2][1]

### 1.4.3.  구체적 성과와 프로그램

1956년, Simon과 Newell, J.C. Shaw는 **Logic Theorist**라는 최초의 인공지능 프로그램을 개발했습니다. 이는 인간의 사고 과정을 모방하여 논리적 문제를 해결할 수 있는 "thinking machine"이었습니다.[3][1][2]

이어서 1957년에는 **General Problem Solver(GPS)**를 개발하여 휴리스틱 기법과 수단-목적 분석을 통해 다양한 문제를 해결하는 시스템을 구축했습니다.[4][1]

### 1.4.4.  조직적 확장

Simon은 Carnegie Mellon에서 제한된 합리성 연구를 위한 학제간 접근을 위해 여러 학과와 기관을 설립하거나 확장하는 데 중요한 역할을 했습니다:[2]

- **컴퓨터 과학과** (1965년 설립)
- **심리학과** 창설/확장
- **사회 및 의사결정 과학과** 창설
- **철학과** 확장
- **비즈니스 스쿨** (현재의 Tepper School of Business) 설립

### 1.4.5.  이론적 기여

Simon은 제한된 합리성을 해결하기 위한 핵심 개념으로 **"satisficing"**(만족화) 전략을 제시했습니다. 이는 최적해를 찾기보다는 충분히 만족할 만한 해를 찾는 것으로, 인간의 인지적 한계 내에서 합리적 의사결정을 하는 방법입니다.[5][6][7]

Simon과 Newell의 이러한 연구는 1975년 **ACM 튜링상** 수상으로 인정받았으며, Simon은 1978년 **노벨경제학상**을 수상했습니다. 이들이 설립한 조직과 개발한 이론은 현재까지도 인공지능과 인지과학 분야의 기초가 되고 있습니다.[8][1][4][2]

## 1.7.머신러닝 주요 구성요소 (용어 정리)
- 데이터(Data):
  - 학습의 가반이 되는 자료.
  - 입력(input)과 정답(Label) 또는 출력(Output)으로 구성됨.
  - 양질의 데이터가 모델의 성능에 결정적인 역할을 함.

- 기계 == 모델(Model):
  - 데이터를 기반으로 학습된 알고리즘(수학적/통계적) 구조. 
  - 입력 데이터를 받아 결과를 출력하는 함수 역할

- 학습(Training): 파블로브의 개, 스키너의 쥐 실험을 통해서 보상시스템이 학습곡선을 향상시킨다는 결과를 가져왔다. 따라서 기계도 동일 Logic을 통해서 학습시킨다.
  - 데이터의 규칙(pattern)을 찾아내는 과정.
  - 예측의 오류(오차)를 최소화하는 방향으로 학습함.

- 예측(Prediction): Statistic Analysis를 통해서 Standard Deviation Table을 만들어 내고, 오차범위(신뢰수준 95%)를 기준으로 Data만 만족해로 사용한다. 
  - 학습된 모델을 사용해 새로운 데이터에 적용하여 결과를 도출하는 과정

- 평가(Evaluation): Null Hypothesis와 Alterative Hypothesis를 통해서 측정값이 얼마나 정확한지를 평가한다.
  - 모델의 예측 성능을 평가하는 과정

# 3.모델의 성능을 이해하고 정리하는 주요 개념
- 과대적합(Over-fitting): 편차가 Over_Value_Bias
  - 모델의 규칙이 너무 복잡해진 상태
  - **훈련**데이터의 **성능**은 높지만 새로운 데이터를 예측했을 때 성능이 떨어지는 상태

- 과소적합(Under-fitting): 편차가 Under_Value_Bias
  - 모델의 규칙이 너무 단순해진 상태
  - **훈련**데이터와 **테스트**데이터 모두 예측 성능이 떨어지는 상태

- 일반화(unbias): 편이가 없다 unbias 
  - 일반적으로 모델의 성능이 높고 보편화됨(좋은 모델)

- 교차검증(Cross Validation): 
  - 모델의 일반화 성능을 확인하기 위해서 여러번 테스트 데이터를 만들고 점수를 확인하는 방법

## 4.1.Supervised Learning에서 Supervised란 무엇을 의미하는지, 지도학습 내에서 상세 유형을 찾아보기
- Supervised란? 
  - 독립변수(X) => Question(특성;Feature Data) => input 
  - 종속변수(y) => Answer(레이블;Label Data)=> output 
  - 특성 데이터(X - 독립변수)

        차원성: 보통 m×n 행렬 형태 (m: 샘플 수, n: 특성 수)

        역할: 예측을 위한 입력 정보로, 결과에 영향을 미치는 요인들

        표기법: 대문자 X를 사용하는 이유는 일반적으로 다차원 행렬이기 때문

        예시:

        주택 가격 예측: 면적, 방 개수, 위치, 건축연도 등

        신용평가: 소득, 직업, 신용기록, 부채비율 등

  - 레이블 데이터(y - 종속변수)

        차원성: 보통 m×1 벡터 형태 (m: 샘플 수)

        역할: 모델이 예측해야 할 목표값

        표기법: 소문자 y를 사용하는 이유는 상대적으로 단순한 구조이기 때문

        특징: 독립변수들의 함수로 표현됨 (y = f(X))
  
  - Classify
  - Regression

## 4.2.Machine Learning에서 E-mail 스팸 분류, 가격 예측이 서로 다른점?

| 유형                                 | 설명                    | 예시               |
| ---------------------------------- | --------------------- | ---------------- |
| **지도 학습 (Supervised Learning)**    | 입력과 정답(Label)을 이용해 학습 | 이메일 스팸 분류, 가격 예측 |


## 5.5.Select of Model 모델 선택(및 하이퍼 파라미터 조정)
- HyperParameter의 궁극적 목적: Bias와 Variance의 균형을 맞추는것 : 편향-분산 균형(Bias-Variance Tradeoff)
    - 편향(Bias): 모델의 예측과 실제 값의 차이
    - 분산(Variance): 새로운 데이터에 대한 모델의 민감도
- 하이퍼 파리미터: 사람이 설정하는 파라미터 값 -> 모델마다 다름
- 모델 선택: 분류 문제를 잘 해결 할 모델 선택


- CF) Parameters vs Hyperparameters: 본질적 차이점
- Parameters (매개변수)
    - 학습 과정에서 자동으로 도출되는 내부 변수
    - 데이터로부터 자동으로 학습되고 업데이트됨
    - 예: 선형회귀의 가중치, 신경망의 연결 강도

- Hyperparameters (하이퍼파라미터)
    - 학습 시작 전에 수동으로 설정하는 외부 구성 변수
    - 모델의 학습 과정 자체를 제어
    - 예: 학습률, 배치 크기, 신경망 층 수

- 이러한 구분은 **"하이퍼(Hyper)"**라는 접두사가 붙는 이유를 설명합니다 - 일반 매개변수보다 상위 수준에서 학습 과정을 통제하기 때문입니다.


```python
# KNN 모델 생성
# 이웃의 수(K-neighbors == n_neighbors)가 하이퍼 파라미터
knn = KNeighborsClassifier(n_neighbors=1) #최적 1
# 이웃 1개만 확인해서 예측하는 설정
```

## 5.6.Learning 모델 학습
- fit(): 데이터에 맞춰서 학습, 데이터연결
- 훈련용: X_train, y_train

```python
# 모델명.fit(훈련문제, 훈련답)
knn.fit(X_train, y_train)
```

## 5.7.Espect&Evaluation모델 예측 및 평가

```python
# test 활용 예측
# X_test, y_test

# 예측을 진행하면서, 답을 알려주는 예측이 있는가?
# 문제를 출제
# 모델명.predict(테스트문제)
pred = knn.predict(X_test)
print("-입력:")
display(X_test)
print("-예측 결과:", pred)
print("-실제값:")
display(y_test)
```

```python
# 정확도 지표 성능확인
from sklearn.metrics import accuracy_score
# accuracy_score(실제값, 예측값)
accuracy_score(y_test,pred) # 1.0 => 100% (백분율) 맞춘 상황
```

```python
# 실제 0.1
# 예측 0,1
# 조합을 표로 정리해놓은 기능 -> 혼동행렬표(cunfusion_matrix)
from sklearn.metrics import confusion_matrix
cf = confusion_matrix(y_test, pred)

cf_df = pd.DataFrame(cf, index = ["실제0", "실제1"],
                     columns = ["예측0", "예측1"])

cf_df

# accurancy 결과를 해석할 수 있는 행렬표
# 실제 0인 것을 예측 0이라고 잘 맞춤 : 1개
# 실제 1인 것을 예측 1이라고 잘 맞춤 : 2개
# accuracy 3/3 = 1.0 100% 맞춘 상황!
```
