- 추가적인 전처리 : 감성점수 전체 문장 적용 전에 KNU 단어사전에 적합한 품사들(동사, 형용사 등)을 추출하여 활용

```python
# okt 형태소 분석기 활용해보자!
from konlpy.tag import Okt
```

```python
# 객체 생성
okt = Okt()

# 어떤 품사 개념을 가지고 있을까? 
print(okt.tagset)

# 품사 선택
# 리뷰내에서 형용사, 동사, 부사, 감탄사 해당하는 단어만 분석에 활용해보자!(해당하는 단어만 필터링) 

def okt_pos_tagging(text):
    # text 받아들여진 문장을 바로 품사 태깅
    pos_words = okt.pos(text, stem = True, norm = True)  # stem : 기본형태로 변환, norm : 정규화(오타 교정)

    # 형용사, 동사, 부사, 감탄사만 추출
    tagged_list = []

    # 반복을 진행하면서 단어를 검사
    for word, tag in pos_words:
        if tag in ['Adjective', 'Verb', 'Adverb', 'Exclamation']:
            tagged_list.append(word)

    #
    return tagged_list
```

```python
# pos_words 내용은 아래와 같음
okt.pos(df["review"].iloc[0])
```

```python
# 함수 2개 정의함
# - 1.형태소 품사를 필터링 : okt_pos_tagging() => 잘라지지 않은 문장을 parameter로 넣어준다 => okt_pos_tagging(리뷰문장) => 토큰화시키기
# - 2.감성 점수 매칭: sentiment_score() => 토큰화된 문장을 parameter로 넣어준다 => sentiment_score(감성사전 ,토큰화된문장) => 우리 고객들만의 감성사전(sentiment)
```

- 모든 리뷰 문장의 감정 점수 계산


```python
# 진행 현황을 파악하기 위한 로딩바 출력
from tqdm.auto import tqdm

# 우리 고객들만의 감성사전 변수 초기화
sentiment = []

for review in tqdm(df["review"]): # 
    okt_pos_tagged = okt_pos_tagging(review) # 
    
    score = sentiment_score(sent_dict, okt_pos_tagged) # 
    sentiment.append(score) # 
```

```python
# list = []
# for i in range(0, len(sentiment[0])):
#     list.append(int(sentiment[0][i][0])) # 감성 점수만 출력

# # 0번째 문장의 감성 점수들

# print(np.mean(list))
```

```python
# 평균 감성 점수를 담을 리스트 초기화(선언)
avg_sentiment_scores = []

for sentiment_result in sentiment:
    # 현재 리뷰의 모든 감성 점수를 정수형으로 변환하는 부분
    #   lst = []
    #   for i in range(0, len(sentiment[0])):
    #  lst.append(int(sentiment[0][i][0])) # 감성 점수만 출력
    # >> [0, 1, 2, 3, 4]
    # for 문 돌리고 list에 최종 값이 저장되는 구조를 : list 내포 문법으로 구현
    # scores = [리스트에 저장될 끝값 for i in sentiment_result ]
    scores = [int(result[0]) for result in sentiment_result] 


    # 평균구하기
    # np.mean(scores)
    # 점수가 있다면 평균을 계산하고, 없다면 0을 저장(분리된 형태소의 단어가 감성사전에 없는 경우)
    avg_score = np.mean(scores) if scores else 0

    # 최종 리스트에 저장
    avg_sentiment_scores.append(avg_score)
```

### 1.1.4. Actor에 대한 Action별 감성점수 계산 후 정규화(스케일링, 데이터 범위 조정) => 이해하기 쉬운형식으로 만들기 => 만족도 점수 산출

```python
# 현재 Actor는 2명(.data/data frame이 2) => 2명의 Actor에 대한 감성 점수 계산
df.groupby(["cluster", "action_cluster"]).count()
```

- dict 형식으로 변환

```python
df["cluster"].unique()
```

```python
# actor의 각 action(행동) 클러스터별 평균 감성 점수를 저장할 딕셔너리 정의 
action_sentiments = {}

for actor in df["cluster"].unique(): # array([0, 1])
    # actor 별로 df 할당 - 인덱싱
    # 0번 actor의 정보만 접근해보자
    # 1번 actor의 정보만 접근해보자
    actor_df = df[df["cluster"] == actor]

    # 각 actor의 action 별로 평균 감성 점수 계산
    for i in actor_df["action_cluster"].unique():
        # 현재 액션 클러스터에 해당되는 행들의 감성점수(sentiment_score) 접근
        action_score = actor_df[actor_df["action_cluster"] == i] ["sentiment_score"]

        # 선택된 액션 클러스터의 평균 감성 점수 계산
        action_score_avg = np.mean(action_score)

        # action_score_avg 저장하기 => action_sentiments
        action_sentiments[f"Actor{actor}_Action{i}"] = action_score_avg
```

```python
# 스케일링 된 점수를 기존 딕셔너리에 업데이트

# 중요도 점수를 키-값 형태로 보관하기 위해 빈 딕셔너리 생성
# - key: "Actor{actor}_Action{action}" 같은 레이블
# - value: MinMaxScaler로 1~10 구간으로 정규화된 중요도 점수
# 리스트([])로 만들면 문자열 키로 할당할 수 없으니 반드시 dict({})를 사용
importance_dict = {}

# zip으로 두 시퀀스를 쌍(pair)으로 묶어 순회
# - label_frequency.keys(): 각 카테고리 레이블(삽입 순서 유지)
# - scaled_scores_list_round: 정규화된 중요도 점수 목록(소수 4자리 반올림)
# 주의: list(...)로 감싸는 것은 불필요하며 메모리만 더 씁니다. 그냥 zip(...)을 사용해도 동일하게 동작함.
for key, new_score in list(zip(label_frequency.keys(), scaled_scores_list_round)):
    # 딕셔너리에 매핑 저장. 존재하는 키면 덮어쓰고, 없으면 신규 추가
    importance_dict[key] = new_score

# Jupyter에서는 셀의 마지막 표현식이 출력되므로, 현재 딕셔너리 내용을 화면에 확인 가능
# 이후 DataFrame에 넣을 때는 순서에 의존하지 말고 'Action' 컬럼을 기준으로 map 하는 것이 안전:
# action_summary_df["importance"] = action_summary_df["Action"].map(importance_dict)
importance_dict
```

### 1.3.1 Opportunity 점수 계산
- Opportunity = Importance + max(Importance-Satisfaction,0)
- 기회 점수는 고객이 어떤 행동을 중요하게 여기면서 동시에 만족도가 낮을 경우, 그 행동에서 개선할 여지가 많음을 나타내는 지표
- 중요도 : 고객이 특정 행동에 대해 얼마나 중요한지를 나타내는 점수
- 만족도 : 고객이 실제로 그 행동에 대해 느끼는 만족도를 나타내는 점수
- 중요도 - 만족도 : 만약 고객이 어떤 행동을 중요하게 여기지만 그에 비해 만족도가 낮다면, 그 차이가 클수록 고객에게 개선의 여지가 많다는 뜻
- max(중요도-만족도,0) : 중요도가 만족도보다 높을 때 그 차이를 사용하고, 그렇지 않을 경우에는 0으로 처리(고객의 만족도가 중요도보다 높으면 추가 점수를 주지 않음)
- 기회 점수 계산 : 기본적으로 고객이 행동에 부여한 중요도에, 중요도와 만족도의 차이가 양수(즉, 개선할 여지)가 있을 때 그 차이를 더해줌
 

```python
# 기회점수 
def opportunity_score(sat, imp):
    return imp + max(imp - sat, 0) # 만약 importance - satisfaction 이 0이라면 0으로 출력하도록 함
```

```python
# 기회점수 계산
opportunity_list = []
for s, i in zip(action_summary_df["Satisfaction"], action_summary_df["importance"]):
    result = opportunity_score(sat=s, imp=i)
    opportunity_list.append(result)
```

```python
# action_summary_df 컬럼 저장
action_summary_df["opportunity_score"] = opportunity_list
action_summary_df

# 저장 
action_summary_df.to_csv("./data/sent_impo_oppo_score.csv", index=False) # index= False : 인덱스를 저장하지 않음. 
```
