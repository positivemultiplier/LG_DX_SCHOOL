## 2.개념적 모델링(청사진, Blueprint)
- 
- 청사진(Blueprint)
- 현실 세계의 복잡한 대상들을 추상화, 단수화 하여 데이터로 표현하는 과정
- 어떤 데이터를 저장할 것인지 결정하는 단계
- ex) 회원가입시 '나'라는 사용자를 네이버가 어떻게 데이터로 저장하는가?

### 2.1. Entity(개체) => table명
- 1. 복잡한 대상들을 하나의 집합으로 표현
- ex) 회원, 블로그, 상품, 주문, 게시글 등등

### 2.2. Attribute(속성) => column
- 해당 집합에서 공통적인 특징들을 표현 
- ex) 이름, ID, 비밀번호, 생년월일, 성별 등등

### 2.3. Instance(인스턴스) => row 각 객체가 실제가 되는것
- 개체의 속성에 실제로 저장된 값
- ex) 홍길동, abc123, qwer1234, 1990-01-01, 남자 등등

### 2.4. Relation(관계) 개념적 모델링에서 관계를 만들어내는것이 가장 중요하다. 
- 관계 : 개체와 개체가 맺고 있는 의미 있는 연관성
- ex) 회원은 블로그를 작성하는 관계
- ex) 블로그는 회원에게 작성되는 관계

### 3.1.식별자(Primary Key) 지정 => 유일성(중복되면 안됨, NaN 안됨)
- 
- 하나의 엔티티의 속성중에서 각 인스턴트들을 식별(indiffence)할 수 있는 속성
    - 성별 (중복되므로 X), 이름(중복되므로 X), ID(유일하므로 O), 주민번호(유일하므로 O)
    - 주민번호는 식별자 후보가 될 수 있으나 적절하지 않을 수 있다. 

### 3.2. 정규화(Normalization) => 중복제거 => 무결성 강화
- 데이터의 중복을 제거하고, 데이터 간의 종속성을 최소화하는 과정
- 
- 회원 entity를 쪼갠다. => 회원ID , pw, 이름, 성별, 나이 // 연락처, 주소 쪼갬. 
- 연락처// 주소 쪼갬 => 3가지의 entity로 나눈다 .
