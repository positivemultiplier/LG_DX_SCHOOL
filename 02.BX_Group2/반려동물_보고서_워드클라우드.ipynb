## 2. 라이브러리 임포트

필요한 모든 라이브러리를 임포트합니다.

```python
import pdfplumber
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from konlpy.tag import Okt
import re
import os
from collections import Counter

# 한글 폰트 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

print("라이브러리 임포트 완료!")
```

## 3. PDF에서 텍스트 추출

pdfplumber를 사용하여 PDF 파일에서 텍스트를 추출합니다.

```python
# PDF 파일 경로
pdf_path = r"c:\Users\lgdx\LG_DX_School\02_DX_Methodology\2025 한국 반려동물 보고서.pdf"

# PDF에서 텍스트 추출
text_content = ""

try:
    with pdfplumber.open(pdf_path) as pdf:
        print(f"총 페이지 수: {len(pdf.pages)}")
        
        for i, page in enumerate(pdf.pages):
            page_text = page.extract_text()
            if page_text:
                text_content += page_text + "\n"
            print(f"페이지 {i+1} 처리 완료")
    
    print(f"\n추출된 텍스트 길이: {len(text_content)} 문자")
    print("\n첫 500자:")
    print(text_content[:500])
    
except Exception as e:
    print(f"PDF 읽기 오류: {e}")
```

## 4. 한국어 텍스트 전처리

추출된 텍스트를 정제하고 형태소 분석을 통해 워드클라우드에 적합한 형태로 가공합니다.

```python
# 불용어 정의 (반려동물 보고서에 특화된 불용어)
stopwords = {
    '것', '그', '이', '저', '수', '년', '월', '일', '대', '등', '및', '한', '등등', 
    '또한', '하는', '있는', '되는', '그리고', '하지만', '때문에', '통해', '위해',
    '같은', '다른', '많은', '높은', '낮은', '크게', '작게', '매우', '상당히',
    '전체', '일부', '부분', '경우', '시', '때', '중', '내', '외', '이상', '이하',
    '약', '정도', '가장', '더', '덜', '보다', '만큼', '처럼', '듯이', '같이',
    '그런', '이런', '그것', '이것', '저것', '어떤', '어떻게', '왜', '언제', '어디',
    '누구', '무엇', '어느', '의한', '에게', '에서', '으로', '로서', '에는', '에도',
    '단위', '가구', '마리', '명', '개', '자료', '출처', '조사', '결과', '그림',
    '표', '페이지', '번호', '차례', '목차', '내용'
}

def preprocess_text(text):
    """텍스트 전처리 함수 (Java 없이 간단한 한글 처리)"""
    # 특수문자 및 숫자 제거 (한글, 영문, 공백만 유지)
    text = re.sub(r'[^\w\s가-힣]', ' ', text)
    text = re.sub(r'\d+', '', text)
    
    # 줄바꿈과 탭을 공백으로 변경
    text = re.sub(r'[\n\t\r]', ' ', text)
    
    # 여러 공백을 하나로
    text = re.sub(r'\s+', ' ', text)
    
    # 간단한 한글 단어 추출 (2글자 이상 한글 단어)
    words = re.findall(r'[가-힣]{2,}', text)
    
    # 불용어 제거
    filtered_words = [
        word for word in words 
        if word not in stopwords and len(word) >= 2
    ]
    
    return filtered_words

# 텍스트 전처리 실행
print("텍스트 전처리 중...")
processed_words = preprocess_text(text_content)

# 단어 빈도 계산
word_freq = Counter(processed_words)

print(f"추출된 단어 수: {len(processed_words)}")
print(f"고유 단어 수: {len(word_freq)}")
print("\n가장 빈도가 높은 상위 20개 단어:")
for word, freq in word_freq.most_common(20):
    print(f"{word}: {freq}")
```

## 5. 워드클라우드 생성

전처리된 텍스트 데이터를 바탕으로 워드클라우드를 생성합니다.

```python
# 워드클라우드를 위한 텍스트 문자열 생성
wordcloud_text = ' '.join(processed_words)

# 한글 폰트 경로 설정 (Windows 시스템의 기본 한글 폰트)
font_path = 'C:/Windows/Fonts/malgun.ttf'  # 맑은 고딕

# 워드클라우드 생성
wordcloud = WordCloud(
    font_path=font_path,
    width=1200,
    height=800,
    background_color='white',
    max_words=100,
    colormap='viridis',
    relative_scaling=0.5,
    random_state=42
).generate_from_frequencies(word_freq)

# 워드클라우드 시각화
plt.figure(figsize=(15, 10))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('2025 한국 반려동물 보고서 워드클라우드', fontsize=20, pad=20)
plt.tight_layout()
plt.show()

print("워드클라우드 생성 완료!")
```

## 6. 워드클라우드 커스터마이제이션

다양한 스타일과 색상으로 워드클라우드를 커스터마이즈합니다.

```python
# 다양한 컬러맵으로 워드클라우드 생성
color_schemes = ['viridis', 'plasma', 'inferno', 'magma', 'Blues', 'Reds']

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('2025 한국 반려동물 보고서 - 다양한 스타일의 워드클라우드', fontsize=16)

for i, colormap in enumerate(color_schemes):
    row = i // 3
    col = i % 3
    
    # 각 컬러맵으로 워드클라우드 생성
    wc = WordCloud(
        font_path=font_path,
        width=600,
        height=400,
        background_color='white',
        max_words=80,
        colormap=colormap,
        relative_scaling=0.5,
        random_state=42
    ).generate_from_frequencies(word_freq)
    
    axes[row, col].imshow(wc, interpolation='bilinear')
    axes[row, col].set_title(f'컬러맵: {colormap}', fontsize=12)
    axes[row, col].axis('off')

plt.tight_layout()
plt.show()
```

## 7. 분석 결과 요약 및 저장

워드클라우드 분석 결과를 요약하고 이미지 파일로 저장합니다.

```python
# 최종 워드클라우드 생성 및 저장
final_wordcloud = WordCloud(
    font_path=font_path,
    width=1600,
    height=1000,
    background_color='white',
    max_words=150,
    colormap='viridis',
    relative_scaling=0.5,
    random_state=42
).generate_from_frequencies(word_freq)

# 시각화
plt.figure(figsize=(20, 12))
plt.imshow(final_wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('2025 한국 반려동물 보고서 워드클라우드 - 최종버전', fontsize=24, pad=30)

# 이미지 저장
output_path = r"c:\Users\lgdx\LG_DX_School\02_DX_Methodology\반려동물_보고서_워드클라우드.png"
plt.savefig(output_path, dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print(f"워드클라우드가 다음 경로에 저장되었습니다: {output_path}")

# 분석 결과 요약
print("\n=== 분석 결과 요약 ===")
print(f"• 총 추출된 단어 수: {len(processed_words):,}개")
print(f"• 고유 단어 수: {len(word_freq):,}개")
print(f"• 평균 단어 길이: {sum(len(word) for word in processed_words) / len(processed_words):.1f}글자")

print("\n=== 주요 키워드 TOP 15 ===")
for i, (word, freq) in enumerate(word_freq.most_common(15), 1):
    print(f"{i:2d}. {word:10s} ({freq:3d}회)")

# 카테고리별 분석
pet_related = [word for word, freq in word_freq.items() if any(keyword in word for keyword in ['반려', '애완', '동물', '개', '고양이', '강아지'])]
market_related = [word for word, freq in word_freq.items() if any(keyword in word for keyword in ['시장', '산업', '업계', '매출', '규모'])]

print(f"\n• 반려동물 관련 키워드: {len(pet_related)}개")
print(f"• 시장/산업 관련 키워드: {len(market_related)}개")
```
